```
import os, sys, yaml, numpy as np, torch, torch.nn as nn, cv2, glob
from torch.utils.data import Dataset, DataLoader
from PIL import Image
from tqdm import tqdm
import segmentation_models_pytorch as smp

# ── Config ─────────────────────────────────────────────────────
DATASET_PATH = "/content/drive/MyDrive/VDD_project/VDD/VDD"
IMAGE_SIZE   = 512
BATCH_SIZE   = 4
EPOCHS       = 20
LR           = 0.0001
NUM_CLASSES  = 7
CHECKPOINT   = "/content/drive/MyDrive/VDD_project/efficientnet_clean_best.pth"

# ── Corruptions ─────────────────────────────────────────────────
def add_fog(image, severity=2):
    fog_intensity = [0.3, 0.5, 0.7, 0.85, 0.95][severity - 1]
    fog_layer = np.ones_like(image, dtype=np.float32) * 255
    result = image.astype(np.float32) * (1 - fog_intensity) + fog_layer * fog_intensity
    return np.clip(result, 0, 255).astype(np.uint8)

def add_gaussian_noise(image, severity=2):
    std = [0.04, 0.06, 0.09, 0.17, 0.26][severity - 1] * 255
    noise = np.random.normal(0, std, image.shape).astype(np.float32)
    return np.clip(image.astype(np.float32) + noise, 0, 255).astype(np.uint8)

def add_blur(image, severity=2):
    k = [3, 5, 7, 9, 11][severity - 1]
    return cv2.GaussianBlur(image, (k, k), 0)

def reduce_brightness(image, severity=2):
    factor = [0.8, 0.6, 0.45, 0.3, 0.15][severity - 1]
    return np.clip(image.astype(np.float32) * factor, 0, 255).astype(np.uint8)

def reduce_contrast(image, severity=2):
    factor = [0.8, 0.6, 0.4, 0.2, 0.1][severity - 1]
    mean = np.mean(image)
    return np.clip((image.astype(np.float32) - mean) * factor + mean, 0, 255).astype(np.uint8)

CORRUPTIONS = {
    "fog": add_fog,
    "gaussian_noise": add_gaussian_noise,
    "blur": add_blur,
    "brightness": reduce_brightness,
    "contrast": reduce_contrast,
}

# ── Dataset ─────────────────────────────────────────────────────
class VDDDataset(Dataset):
    def __init__(self, dataset_path, split="train", image_size=512, augmentation=False, corruption=None, severity=2):
        self.image_size  = image_size
        self.augmentation = augmentation
        self.corruption  = corruption
        self.severity    = severity
        self.image_dir   = os.path.join(dataset_path, split, "src")
        self.mask_dir    = os.path.join(dataset_path, split, "gt")
        self.images      = sorted(glob.glob(os.path.join(self.image_dir, "*.JPG")))
        print(f"[{split}] Found {len(self.images)} images")

    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):
        img_path  = self.images[idx]
        name      = os.path.splitext(os.path.basename(img_path))[0]
        mask_path = os.path.join(self.mask_dir, name + ".png")
        image = cv2.cvtColor(cv2.imread(img_path), cv2.COLOR_BGR2RGB)
        mask  = np.array(Image.open(mask_path))
        if self.augmentation:
            corruption = np.random.choice(list(CORRUPTIONS.keys()))
            image = CORRUPTIONS[corruption](image, severity=np.random.randint(1, 4))
        if self.corruption:
            image = CORRUPTIONS[self.corruption](image, self.severity)
        image = cv2.resize(image, (self.image_size, self.image_size))
        mask  = cv2.resize(mask,  (self.image_size, self.image_size), interpolation=cv2.INTER_NEAREST)
        mask  = np.clip(mask, 0, 6)
        image = torch.from_numpy(image.astype(np.float32) / 255.0).permute(2, 0, 1)
        mask  = torch.from_numpy(mask).long()
        return image, mask

# ── Metric ──────────────────────────────────────────────────────
def compute_miou(preds, labels, num_classes=7):
    preds, labels = preds.cpu().numpy(), labels.cpu().numpy()
    ious = []
    for cls in range(num_classes):
        inter = ((preds == cls) & (labels == cls)).sum()
        union = ((preds == cls) | (labels == cls)).sum()
        if union > 0:
            ious.append(inter / union)
    return np.mean(ious) if ious else 0.0

# ── Model ───────────────────────────────────────────────────────
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

model = smp.DeepLabV3Plus(
    encoder_name    = "efficientnet-b0",
    encoder_weights = "imagenet",
    in_channels     = 3,
    classes         = NUM_CLASSES,
).to(device)

optimizer = torch.optim.Adam(model.parameters(), lr=LR)
criterion = nn.CrossEntropyLoss()

# ── DataLoaders ─────────────────────────────────────────────────
train_dl = DataLoader(VDDDataset(DATASET_PATH, "train", IMAGE_SIZE, augmentation=False),
                      batch_size=BATCH_SIZE, shuffle=True,  num_workers=2)
val_dl   = DataLoader(VDDDataset(DATASET_PATH, "val",   IMAGE_SIZE, augmentation=False),
                      batch_size=BATCH_SIZE, shuffle=False, num_workers=2)

# ── Training Loop ───────────────────────────────────────────────
best_miou = 0.0
for epoch in range(1, EPOCHS + 1):
    model.train()
    total_loss = 0
    for images, masks in tqdm(train_dl, desc=f"Epoch {epoch}/{EPOCHS} [Train]"):
        images, masks = images.to(device), masks.to(device)
        optimizer.zero_grad()
        loss = criterion(model(images), masks)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()

    model.eval()
    miou_scores = []
    with torch.no_grad():
        for images, masks in tqdm(val_dl, desc=f"Epoch {epoch}/{EPOCHS} [Val]"):
            images, masks = images.to(device), masks.to(device)
            preds = torch.argmax(model(images), dim=1)
            miou_scores.append(compute_miou(preds, masks))

    avg_miou = np.mean(miou_scores)
    avg_loss = total_loss / len(train_dl)
    print(f"Epoch {epoch} | Loss: {avg_loss:.4f} | Val mIoU: {avg_miou:.4f}")

    if avg_miou > best_miou:
        best_miou = avg_miou
        torch.save(model.state_dict(), CHECKPOINT)
        print(f"  Saved best model! mIoU: {best_miou:.4f}")

print(f"\nTraining complete! Best mIoU: {best_miou:.4f}")
```

```
from google.colab import drive
drive.mount('/gdrive')

import os, sys, yaml, numpy as np, torch, torch.nn as nn, cv2, glob
from torch.utils.data import Dataset, DataLoader
from PIL import Image
from tqdm import tqdm
import segmentation_models_pytorch as smp

# ── Config ─────────────────────────────────────────────────────
DATASET_PATH = "/gdrive/MyDrive/VDD_project/VDD/VDD"
IMAGE_SIZE   = 512
BATCH_SIZE   = 4
EPOCHS       = 20
LR           = 0.0001
NUM_CLASSES  = 7
CHECKPOINT   = "/gdrive/MyDrive/VDD_project/efficientnet_clean_best.pth"

# ── Corruptions ─────────────────────────────────────────────────
def add_fog(image, severity=2):
    fog_intensity = [0.3, 0.5, 0.7, 0.85, 0.95][severity - 1]
    fog_layer = np.ones_like(image, dtype=np.float32) * 255
    result = image.astype(np.float32) * (1 - fog_intensity) + fog_layer * fog_intensity
    return np.clip(result, 0, 255).astype(np.uint8)

def add_gaussian_noise(image, severity=2):
    std = [0.04, 0.06, 0.09, 0.17, 0.26][severity - 1] * 255
    noise = np.random.normal(0, std, image.shape).astype(np.float32)
    return np.clip(image.astype(np.float32) + noise, 0, 255).astype(np.uint8)

def add_blur(image, severity=2):
    k = [3, 5, 7, 9, 11][severity - 1]
    return cv2.GaussianBlur(image, (k, k), 0)

def reduce_brightness(image, severity=2):
    factor = [0.8, 0.6, 0.45, 0.3, 0.15][severity - 1]
    return np.clip(image.astype(np.float32) * factor, 0, 255).astype(np.uint8)

def reduce_contrast(image, severity=2):
    factor = [0.8, 0.6, 0.4, 0.2, 0.1][severity - 1]
    mean = np.mean(image)
    return np.clip((image.astype(np.float32) - mean) * factor + mean, 0, 255).astype(np.uint8)

CORRUPTIONS = {
    "fog": add_fog,
    "gaussian_noise": add_gaussian_noise,
    "blur": add_blur,
    "brightness": reduce_brightness,
    "contrast": reduce_contrast,
}

# ── Dataset ─────────────────────────────────────────────────────
class VDDDataset(Dataset):
    def __init__(self, dataset_path, split="train", image_size=512, augmentation=False, corruption=None, severity=2):
        self.image_size   = image_size
        self.augmentation = augmentation
        self.corruption   = corruption
        self.severity     = severity
        self.image_dir    = os.path.join(dataset_path, split, "src")
        self.mask_dir     = os.path.join(dataset_path, split, "gt")
        self.images       = sorted(
            glob.glob(os.path.join(self.image_dir, "*.JPG")) +
            glob.glob(os.path.join(self.image_dir, "*.jpg"))
        )
        print(f"[{split}] Found {len(self.images)} images")

    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):
        img_path  = self.images[idx]
        name      = os.path.splitext(os.path.basename(img_path))[0]
        mask_path = os.path.join(self.mask_dir, name + ".png")
        image = cv2.cvtColor(cv2.imread(img_path), cv2.COLOR_BGR2RGB)
        mask  = np.array(Image.open(mask_path))
        if self.augmentation:
            corruption = np.random.choice(list(CORRUPTIONS.keys()))
            image = CORRUPTIONS[corruption](image, severity=np.random.randint(1, 4))
        if self.corruption:
            image = CORRUPTIONS[self.corruption](image, self.severity)
        image = cv2.resize(image, (self.image_size, self.image_size))
        mask  = cv2.resize(mask,  (self.image_size, self.image_size), interpolation=cv2.INTER_NEAREST)
        mask  = np.clip(mask, 0, 6)
        image = torch.from_numpy(image.astype(np.float32) / 255.0).permute(2, 0, 1)
        mask  = torch.from_numpy(mask).long()
        return image, mask

# ── Metric ──────────────────────────────────────────────────────
def compute_miou(preds, labels, num_classes=7):
    preds, labels = preds.cpu().numpy(), labels.cpu().numpy()
    ious = []
    for cls in range(num_classes):
        inter = ((preds == cls) & (labels == cls)).sum()
        union = ((preds == cls) | (labels == cls)).sum()
        if union > 0:
            ious.append(inter / union)
    return np.mean(ious) if ious else 0.0

# ── Model ───────────────────────────────────────────────────────
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

model = smp.DeepLabV3Plus(
    encoder_name    = "efficientnet-b0",
    encoder_weights = "imagenet",
    in_channels     = 3,
    classes         = NUM_CLASSES,
).to(device)

optimizer = torch.optim.Adam(model.parameters(), lr=LR)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)
criterion = nn.CrossEntropyLoss()

# ── DataLoaders ─────────────────────────────────────────────────
train_dl = DataLoader(VDDDataset(DATASET_PATH, "train", IMAGE_SIZE, augmentation=True),
                      batch_size=BATCH_SIZE, shuffle=True,  num_workers=2)
val_dl   = DataLoader(VDDDataset(DATASET_PATH, "val",   IMAGE_SIZE, augmentation=False),
                      batch_size=BATCH_SIZE, shuffle=False, num_workers=2)

# ── Training Loop ───────────────────────────────────────────────
best_miou = 0.0
for epoch in range(1, EPOCHS + 1):
    model.train()
    total_loss = 0
    for images, masks in tqdm(train_dl, desc=f"Epoch {epoch}/{EPOCHS} [Train]"):
        images, masks = images.to(device), masks.to(device)
        optimizer.zero_grad()
        loss = criterion(model(images), masks)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()

    model.eval()
    miou_scores = []
    with torch.no_grad():
        for images, masks in tqdm(val_dl, desc=f"Epoch {epoch}/{EPOCHS} [Val]"):
            images, masks = images.to(device), masks.to(device)
            preds = torch.argmax(model(images), dim=1)
            miou_scores.append(compute_miou(preds, masks))

    avg_miou = np.mean(miou_scores)
    avg_loss = total_loss / len(train_dl)
    print(f"Epoch {epoch} | Loss: {avg_loss:.4f} | Val mIoU: {avg_miou:.4f}")
    scheduler.step()

    if avg_miou > best_miou:
        best_miou = avg_miou
        torch.save(model.state_dict(), CHECKPOINT)
        print(f"  Saved best model! mIoU: {best_miou:.4f}")

print(f"\nTraining complete! Best mIoU: {best_miou:.4f}")
```
